Scatter plot ...  it's shows the co-relation 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sct = pd.read_csv("batter.csv")
df= sct.head(50) # We are taking only the first 50 rows of the dataset for analysis
plt.scatter(
            df["avg"], 
            df["strike_rate"], 
            color= "blue", 
            s=50,
            marker= "+"
            ) # Scatter plot of average vs strike rate

plt.title("Average vs Striker Rate graph of top 50 batsmen")
plt.xlabel("Average")
plt.ylabel("Strike Rate")

 
```

```python
# We can aslo plot a line graph using the same data to see the graph,
plt.plot(df["avg"], df["strike_rate"],'o', color= "red") # Line plot of average vs strike rate
# plt.plot is much faster the plt.scatter
```
