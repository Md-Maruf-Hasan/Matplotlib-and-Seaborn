Actual operation from dataset 
in pandas 

```python
dataset = pd.read_csv("sharma-kohli.csv")
#print(dataset )
print(plt.plot(dataset["index"], dataset["RG Sharma"]))
print(plt.plot(dataset["index"], dataset['V Kohli']))
```

Mutliple ploting in a single graph 

```python
plt.plot(dataset["index"], dataset["RG Sharma"],label ="Rohit Sharma", color = "red")
plt.plot(dataset["index"], dataset['V Kohli'],label ="Virat Kohli", color = "blue")
# Here index(season) and RG Sharma(runs) from the dataset 
# And label for the legend function which is indicate the colour of the person .And color is for custom colour.

plt.title("Rohit sharma vs Virat Kohli runs comparisom")

plt.xlabel("Season")
plt.ylabel("Runs")
## Indicating the title according to their axis
plt.legend() 
```

Line style 

```python
plt.plot(
    dataset["index"],
    dataset["RG Sharma"],
    label="Rohit Sharma",
    color="red",
    linestyle="dashed",
    linewidth=3,
    marker="*"
)
plt.plot(
    dataset["index"], 
    dataset['V Kohli'],
    label ="Virat Kohli", 
    color = "blue", 
    linestyle = "dashdot", 
    linewidth = 4, 
    marker ="o")

# And we have an other one dotted 
```

Using limit in the graphs X and Y axis 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
Phone_price= [500, 550, 620, 700, 850, 700]# Numerical data in Y axis
Year = [2015, 2016, 2017, 2018, 2019, 2020] # We will try to write catagorical data in X axis
plt.plot(Year , Phone_price)
plt.xlim(2000, 2030) # X axis limit
plt.ylim(0, 1000) # Y axis limit
plt.grid() # Grid is for the background lines in the graph
plt.show() # Show is for the graph to be displayed
```
